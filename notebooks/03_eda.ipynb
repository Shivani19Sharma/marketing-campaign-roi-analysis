```python
from scipy.stats import ttest_ind

retargeting = df.loc[
    df["retargeting_flag"] == True,
    "conversion_rate" 
]

non_retargeting = df.loc[
    df["retargeting_flag"] == False,
    "conversion_rate"
]

t_stat, p_value = ttest_ind(
    retargeting,
    non_retargeting,
    equal_var = False
)

t_stat, p_value
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    retargeting,
    non_retargeting,
    equal_var = False
)

mean_diff = retargeting.mean() - non_retargeting.mean()

#Welch - Satterthwaite degrees of freedom
n1 = len(retargeting)
n2 = len(non_retargeting)

s1 = retargeting.std()
s2 = non_retargeting.std()

se_diff = ((s1**2 / n1) + (s2**2 / n2)) ** 0.5

df_welch = (
    ((s1**2 / n1) + (s2**2 / n2)) ** 2
    /
    (
        ((s1**2 / n1) ** 2 / (n1 - 1))
        +
        ((s2**2 / n2) ** 2 / (n2 - 1))
    )
)

from scipy.stats import t

t_critical = t.ppf(0.975, df_welch)

margin_of_error = t_critical * se_diff

ci_lower = mean_diff - margin_of_error
ci_upper = mean_diff + margin_of_error

mean_diff, ci_lower, ci_upper
```
