```python
print("Click > Impressions:",(df["clicks"]>df["impressions"]).sum())
print("Conversion > Click:",(df["conversions"]>df["clicks"]).sum())
print("Negative Ad Revenue:",(df["ad_spend"]<0).sum())
print("Negative Revenue:",(df["revenue"]<0).sum())
print("Negative Impressions:",(df["impressions"]<0).sum())
print("Negative Clicks:",(df["clicks"]<0).sum())
print("Negative Conversions:",(df["conversions"]<0).sum())
```

```python
df["CTR_calc"] = (df["clicks"]/df["impressions"])*100
df["CPC_calc"] = df["ad_spend"]/df["clicks"]
df["conversion_rate_calc"] = (df["conversions"]/df["clicks"])*100
df["CPA_calc"] = df["ad_spend"]/df["conversions"]
df["ROAS_calc"] = df["revenue"]/df["ad_spend"]
df["profit_calc"] = df["revenue"] - df["ad_spend"]
```

```python
print("CTR max difference:", (df["CTR_calc"]-df["CTR"]).abs().max())
print("CPC max difference:",(df["CPC_calc"]-df["CPC"]).abs().max())
print("Conversion Rate max differnce:",(df["conversion_rate_calc"]-df["conversion_rate"]).abs().max())
print("CPA max difference:",(df["CPA_calc"]-df["CPA"]).abs().max())
print("ROAS max difference:",(df["ROAS_calc"]-df["ROAS"]).abs().max())
print("Profit max difference:",(df["profit_calc"]-df["profit"]).abs().max())
```
